# SQL JOIN + Pandas Integration Mini Project

## Goal
Combine related tables using SQL JOIN and load the result into a Pandas DataFrame using Python.

```
import sqlite3
import pandas as pd
```

## Step 1: Create SQLite Database Connection

```
conn = sqlite3.connect('internship.db')
cursor = conn.cursor()
```

## Step 2: Create interns Table

```
cursor.execute('''
CREATE TABLE IF NOT EXISTS interns (
    intern_id INTEGER PRIMARY KEY,
    intern_name TEXT,
    track TEXT,
    mentor_id INTEGER
)
''')

conn.commit()
```

## Step 3: Create mentors Table

```
cursor.execute('''
CREATE TABLE IF NOT EXISTS mentors (
    mentor_id INTEGER PRIMARY KEY,
    mentor_name TEXT,
    track TEXT
)
''')

conn.commit()
```

## Step 4: Insert Sample Data

```
# Insert interns data
interns_data = [
    (1, 'Chandan', 'Data Science', 101),
    (2, 'Rahul', 'Machine Learning', 102),
    (3, 'Priya', 'Web Development', 103),
    (4, 'Anjali', 'Data Science', 101)
]

cursor.executemany('INSERT OR REPLACE INTO interns VALUES (?, ?, ?, ?)', interns_data)

# Insert mentors data
mentors_data = [
    (101, 'Dr. Sharma', 'Data Science'),
    (102, 'Mr. Kumar', 'Machine Learning'),
    (103, 'Ms. Reddy', 'Web Development')
]

cursor.executemany('INSERT OR REPLACE INTO mentors VALUES (?, ?, ?)', mentors_data)

conn.commit()
```

## Step 5: INNER JOIN Query
This query combines interns with their assigned mentors based on mentor_id.

```
query = '''
SELECT 
    interns.intern_id,
    interns.intern_name,
    interns.track,
    mentors.mentor_name
FROM interns
INNER JOIN mentors
ON interns.mentor_id = mentors.mentor_id
'''

df = pd.read_sql_query(query, conn)

print(df)
```

## Why INNER JOIN?
INNER JOIN returns only the rows where matching mentor IDs exist in both tables. This helps combine normalized data spread across multiple related tables.

# Final Output
The SQL query result is successfully loaded into a Pandas DataFrame for further analysis and machine learning workflows.
